```python
import sys
sys.path.insert(0, "..")          # lets the notebook import common.py from python_practice/
from common import *              # load_operations, check, OUT, ...
import pandas as pd

df = load_operations()            # reads the dashboard API if it is running, else data_samples/operations_clean.csv
import matplotlib.pyplot as plt
```

```python
daily = df.groupby("record_date")[["revenue", "operational_cost", "profit"]].sum()
by_entity = df.groupby("entity_name")["profit"].sum().sort_values()
cat = df.groupby("category")["revenue"].sum()
fig = plt.figure(figsize=(11, 7), constrained_layout=True)
gs = fig.add_gridspec(2, 3)

ax = fig.add_subplot(gs[0, 0]); ax.axis("off")                       # 1. top-left: the headline KPI
ax.text(0, 0.7, f"${df['revenue'].sum():,.0f}", fontsize=26, fontweight="bold"); ax.text(0, 0.45, "total revenue", color="#64748b")
ax.text(0, 0.15, f"margin {df['profit'].sum() / df['revenue'].sum():.1%}", fontsize=13)

ax = fig.add_subplot(gs[0, 1:])                                       # 2. top-right: trend over time -> line
daily["revenue"].rolling(7).mean().plot(ax=ax, color="#2563eb"); ax.set_title("Revenue, 7-day average"); ax.set_xlabel("")

ax = fig.add_subplot(gs[1, :2])                                       # 3. bottom-left: compare categories -> sorted bars
by_entity.plot.barh(ax=ax, color="#059669"); ax.set_title("Profit by entity"); ax.set_ylabel("")

ax = fig.add_subplot(gs[1, 2])                                        # 4. bottom-right: parts of a whole (few slices)
ax.pie(cat, labels=cat.index, autopct="%1.0f%%", colors=["#2563eb", "#d97706", "#7c3aed"][:len(cat)]); ax.set_title("Revenue mix")
plt.show()
```
